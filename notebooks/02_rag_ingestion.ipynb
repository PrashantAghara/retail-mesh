```python
from langchain_postgres import PGVector

DATABASE_URL = os.getenv("DATABASE_URL").replace(
    "postgresql://", "postgresql+psycopg2://"
)
COLLECTION_NAME = "retailmesh_knowledge_base"

all_documents = documents + bitext_documents + new_product_documents
print(f"Total corpus size: {len(all_documents)} documents")

vectorstore = PGVector.from_documents(
    documents=all_documents,
    embedding=embedder,
    collection_name=COLLECTION_NAME,
    connection=DATABASE_URL,
    pre_delete_collection=True,
)
print(
    f"Ingested {len(all_documents)} documents into '{COLLECTION_NAME}' (clean, offensive-content-free)"
)
```

```python
import re
from langchain_community.retrievers import BM25Retriever
from langchain_classic.retrievers import EnsembleRetriever

STOPWORDS = {
    "a",
    "an",
    "the",
    "is",
    "are",
    "do",
    "does",
    "my",
    "your",
    "i",
    "you",
    "it",
    "this",
    "that",
    "of",
    "for",
    "to",
    "in",
    "on",
    "at",
    "and",
    "or",
    "how",
}


def simple_tokenize(text: str):
    tokens = re.findall(r"\b\w+\b", text.lower())
    return [t for t in tokens if t not in STOPWORDS]


bm25_retriever = BM25Retriever.from_documents(
    all_documents, preprocess_func=simple_tokenize
)
bm25_retriever.k = 4

vector_retriever = vectorstore.as_retriever(search_kwargs={"k": 4})

ensemble_retriever = EnsembleRetriever(
    retrievers=[bm25_retriever, vector_retriever],
    weights=[0.3, 0.7],
)
```

```python
from sentence_transformers import CrossEncoder

reranker = CrossEncoder("cross-encoder/ms-marco-MiniLM-L-6-v2")


def hybrid_search_reranked(query: str, top_k: int = 2):
    candidates = ensemble_retriever.invoke(query)
    pairs = [[query, doc.page_content] for doc in candidates]
    scores = reranker.predict(pairs)
    ranked = sorted(zip(candidates, scores), key=lambda x: x[1], reverse=True)
    return ranked[:top_k]
```

```python
test_queries = [
    "Do you have Cadbury chocolate?",
    "What is your return policy for bread?",
    "How do I cancel my order?",
    "Is there a warranty on my blender?",
]

all_test_queries = test_queries + [
    "What's your refund policy?",
    "My delivery partner isn't responding",
    "Can I exchange this in store?",
    "Do you have wireless earbuds?",
    "I need a mixer grinder",
]

for q in all_test_queries:
    results = hybrid_search_reranked(q)
    print(f"\nQuery: {q!r}")
    for doc, score in results:
        print(
            f"  -> [{doc.metadata['doc_type']}/{doc.metadata.get('category')}] {doc.metadata['title'][:60]} (score: {score:.3f})"
        )
```

### Retrieve

```python
def retrieve(query: str, top_k: int = 3) -> list[Document]:
    candidates = ensemble_retriever.invoke(query)
    pairs = [[query, doc.page_content] for doc in candidates]
    scores = reranker.predict(pairs)

    ranked = sorted(zip(candidates, scores), key=lambda x: x[1], reverse=True)
    return [doc for doc, score in ranked[:top_k]]
```

```python
results = retrieve("What is your return policy?")
for doc in results:
    print(
        f"[{doc.metadata['doc_type']}/{doc.metadata.get('category')}] {doc.metadata['title']}"
    )
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_groq import ChatGroq

support_llm = ChatGroq(api_key=GROQ_API_KEY, model="openai/gpt-oss-120b")

support_prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            """You are RetailMesh's customer support assistant. Answer the customer's
question using ONLY the context provided below. If the context doesn't contain
enough information to answer confidently, say so honestly instead of guessing.

Keep answers concise and friendly. Don't mention "the context" or "the documents"
in your reply — answer as if you simply know this information.

Context:
{context}""",
        ),
        ("human", "{query}"),
    ]
)

support_chain = support_prompt | support_llm
```

```python
def support_agent(query: str) -> dict:
    docs = retrieve(query, top_k=3)

    context = "\n\n".join(
        f"[{doc.metadata.get('doc_type')}] {doc.page_content}" for doc in docs
    )

    response = support_chain.invoke({"context": context, "query": query})

    return {
        "query": query,
        "answer": response.content,
        "sources": [
            {
                "title": doc.metadata.get("title"),
                "category": doc.metadata.get("category"),
            }
            for doc in docs
        ],
    }
```

```python
support_test_queries = [
    "What is your return policy for bread?",
    "How do I cancel my order?",
    "Do you have Cadbury chocolate in stock?",
    "My delivery partner isn't responding, what should I do?",
    "Do you sell wireless earbuds?",
]

for q in support_test_queries:
    result = support_agent(q)
    print(f"\nQ: {result['query']}")
    print(f"A: {result['answer']}")
    print(f"Sources: {[s['title'] for s in result['sources']]}")
```
